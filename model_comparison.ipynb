```python
import pandas as pd
from scipy.stats import ks_2samp, wasserstein_distance

real = pd.read_csv('data/synthetic/baseline_auto.csv')
ctgan = pd.read_csv('data/synthetic/ctgan_auto_50k.csv')
tvae = pd.read_csv('data/synthetic/tvae_auto_50k.csv')
```

```python
print("Real rows:", len(real))
print("CTGAN rows:", len(ctgan))
print("TVAE rows:", len(tvae))
```

```python
for col in ['premium', 'claim_amount']:

    print("\nColumn:", col)

    ks_ctgan = ks_2samp(real[col], ctgan[col])
    ks_tvae = ks_2samp(real[col], tvae[col])

    wd_ctgan = wasserstein_distance(real[col], ctgan[col])
    wd_tvae = wasserstein_distance(real[col], tvae[col])

    print("CTGAN KS p-value:", ks_ctgan.pvalue)
    print("TVAE KS p-value:", ks_tvae.pvalue)

    print("CTGAN Wasserstein:", wd_ctgan)
    print("TVAE Wasserstein:", wd_tvae)
```
